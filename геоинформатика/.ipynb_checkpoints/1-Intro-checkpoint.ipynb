###### Python 

Python - широко используемый язык программирования общего назначения высокого уровня. Отметим, что Гвидо ван Россум – первый доброжелательный диктатор на всю жизнь (Benevolent dictator for life – BDFL). Доброжелательный диктатор на всю жизнь —титул, присваиваемый небольшому количеству лидеров разработки программного обеспечения с открытым исходным кодом, обычно основателям проектов, за которыми остается последнее слово в спорах внутри сообщества. Ван Россум объявил 12 июля 2018 года, что уйдет с поста BDFL Python без назначения преемника, чем фактически устранил титул в структуре сообщества Python (некоммерческой организации Python Software Foundation).

Итак, Python был концептуализирован в конце 1980-х годов. Гвидо ван Россум признает, что опыт работы разработчиком ABC в CWI во многом повлиял на дизайн Python: «Я вспомнил весь свой опыт и некоторое разочарование в ABC. Я решил попробовать разработать простой язык сценариев, который обладал бы некоторыми из лучших свойств ABC, но без проблем. Итак, я начал печатать. Я создал простую виртуальную машину, простой синтаксический анализатор и простую среду выполнения. Я создал свою собственную версию различных частей ABC, которые мне нравились. Я создал базовый синтаксис, использовал отступы для группировки операторов вместо фигурных скобок или блоков begin-end и разработал небольшое количество мощных типов данных: хеш-таблицу (или словарь, как мы ее называем), список, строки и числа».

Вдохновением для названия послужило телешоу BBC «Летающий цирк Монти Пайтона», поскольку он был большим поклонником телешоу, а также хотел короткое, уникальное и немного загадочное название для своего изобретения: «Я выбрал Python в качестве рабочего названия для проекта, находясь в слегка хулиганском настроении (и являясь большим поклонником Летающего цирка Монти Пайтона)».

Гвидо Ван Россум опубликовал первую версию кода Python (версия 0.9.0) в феврале 1991 года. Этот выпуск уже включал обработку исключений, функции и основные типы данных, такие как `list`, `dict` и `str`. Он также был объектно-ориентированным и имел модульную систему.

###### Jupyter Notebook

В рамках данного курса все примеры и задания будут использовать интерактивную оболочку IPython, входящую в состав дистрибутива Anaconda. Установить его не сложно. Для этого нужно скачать установочный файл с сайта anaconda.com (на момент написания - https://www.anaconda.com/products/individual#Downloads), затем запустить установку и следовать инструкциям. После успешной установки вы можете запустить IPython с помощью команды в терминале или ярлыка в панели, в зависимости от используемой вами операционной системы, в результате чего в браузере должна открыться новая вкладка с домашней страницей Jupyter Notebook, где вы увидите список файлов рабочего каталога. 

Сразу определимся с терминологией. Ноутбук (или блокнот) — это документ, созданный приложением Jupyter Notebook, который может содержать как исполняемый код, так и элементы форматированного текста (уравнения, рисунки, ссылки и т.д.). Ноутбуки — это как удобочитаемые документы, содержащие описание анализа и его результаты (рисунки, таблицы и т.д.), так и исполняемые документы, которые можно запускать для анализа данных.

Итак, чтобы создать новый ноутбук, воспользуйтесь кнопкой «New» в правом верхнем углу домашней страницы Jupyter Notebook и выберите в выпадающем меню пункт «Python». Вы можете обратить внимание, что вам также доступны варианты создания текстовых документов, папок и терминалов.

После создания ноутбука, автоматически откроется новая вкладка с ним, а в директории, где вы находились, когда нажали «New», появился новый .ipynb-файл – это и есть ваш ноутбук. Вы можете его переименовать (по умолчанию новые ноутбуки будут именоваться «Untitled#», где вместо «#» будет порядковый номер). 

Вкратце рассмотрим основные элементы интерфейса ноутбука. Панель меню в верхней части ноутбука по умолчанию содержит следующие элементы: 

**File**: действия, связанные с файлами и каталогами. В нем вы можете создать новый блокнот или открыть уже существующий, можете переименовать Блокнот. Наиболее интересным пунктом меню является опция «Save and Checkpoint», которая позволяет вам создавать контрольные точки, к которым вы можете откатиться при необходимости. Также важно добавить, что существует возможность экспортировать ноутбук в различные форматы, например, в Python (.py), LaTeX (.tex), PDF (.pdf) или HTML (.html) с помощью пункта «Download as».

**Edit**: действия, связанные с редактированием ноутбуков. Позволяет вам вырезать, копировать и вставлять ячейки, а кроме того, удалить, разделить или объединить ячейку. С помощью данного пункта меню вы также можете изменить порядок ячеек. Обратите внимание, что некоторые элементы в этом меню неактивны. Причина в том, что они не применяются к текущей выбранной ячейке. Например, в ячейку кода нельзя вставить изображение, а в ячейку разметки можно. Если вы видите пункт меню серого цвета, попробуйте изменить тип ячейки и посмотрите, станет ли этот элемент доступным для использования. О типах ячеек вы узнаете чуть позже.

**View**: действия, изменяющие внешний вид ноутбука. Позволяет переключать видимость заголовка и панели инструментов, а также включать или выключать нумерацию строк в ячейках. 

**Insert**: предназначено только для вставки ячеек выше или ниже текущей выбранной ячейки.

**Cell**: позволяет запускать одну ячейку, группу ячеек или все ячейки, а также изменить тип ячейки, однако, панель инструментов для этой цели более интуитивно понятна. Другая удобная функция в этом меню - возможность очистить вывод ячейки (если вы планируете поделиться своим блокнотом с другими, вы, вероятно, захотите сначала очистить вывод, чтобы следующий человек мог запускать ячейки самостоятельно).

**Kernel**: действия для управления ядрами, которые представляют собой отдельные процессы для запуска кода. Позволяет перезапустить ядро, повторно подключиться к нему, выключить его или даже изменить ядро, которое использует ваш ноутбук. Вероятнее всего, вы не будете часто работать с ядрами, но бывают случаи, когда при отладке вам будет необходимо перезапустить ядро.

**Help**: пункт, в котором вы можете узнать о горячих клавишах, ознакомиться с пользовательским интерфейсом и множеством справочных материалов.

В процессе ознакомления с панелью меню, были введены два новых термина. Ядро – «вычислительный движок», который выполняет код, содержащийся в ноутбуке. В рамках данного курса мы будем пользоваться ядром Python 3, но существуют ядра для многих других языков (официальные ядра). Когда вы открываете блокнот, автоматически запускается соответствующее ядро. Когда ноутбук запускается (ячейка за ячейкой или с помощью меню «Cell» -> «Run All»), ядро выполняет вычисления и выдает результаты. В зависимости от типа вычислений ядро может потреблять значительные ресурсы ЦП и ОЗУ. Обратите внимание, что оперативная память не освобождается, пока ядро не будет выключено. И второй термин – ячейка – многострочное текстовое поле ввода, из последовательности которых состоит блокнот. Её содержимое может быть выполнено с помощью Ctrl+Enter (или Shift+Enter или Alt+Enter) или нажатием кнопки «Run» на панели инструментов. Поведение ячейки при выполнении определяется типом ячейки. Есть три типа ячеек: ячейки кода, ячейки разметки и «сырые» ячейки. Каждая ячейка сначала является ячейкой кода, но ее тип можно изменить с помощью раскрывающегося списка на панели инструментов (изначально это будет «Code») или с помощью сочетаний клавиш.
Ниже панели меню расположена уже упомянутая панель инструментов, которая позволяет быстро выполнять наиболее часто используемые операции, щелкая на иконку. Если действие какой-либо из представленных кнопок вам не ясно, то наведите на нее указатель мыши, чтобы увидеть комментарий.

###### Возведение в степень

Во-первых, возведение в степень производится с помощью оператора `**`, а не `^`. Оператор `^` - оператор побитового исключающего ИЛИ, то есть, например, при выполнении `a ^ b` при `a = 0x0101` и `b = 0x1111`, в результате получим `0x1010` (если вы вдруг этого не поняли – не расстраивайтесь – это не потребуется на занятиях, главное – запомните оператор возведения в степень и путайте их).

```python
a = 0b0101
b = 0b1111
bin(a ^ b)
```

```python
a = 2
b = 3
print("a ^ b != a ** b:", a ^ b != a ** b)
print("a ^ b  =", a ^ b)
print("a ** b =", a ** b)
```

###### Деление

Во-вторых, если вы пользуетесь второй версией Python, помните, что при делении целого числа на целое число результатом будет целое число (чтобы получить в результате деления число с плавающей точкой, делимое или делитель нужно привести к числу к плавающей точкой).

В Python 2.х:
> `100 / 35 => 2`
>
> `(100 * 1.0) / 35 => 2.857142857142857`
>
> `100 / (35 * 1.0) => 2.857142857142857`

В Python 3.х:
> `100 / 35 => 2.857142857142857`
>
> `100 // 35 => 2    # // используется для целочисленного деления`
>
> `100.// 35 => 2.0    # результат деления с использованием числа с плавающей точкой - вещественный`

Заголовок будет преобразован в интерактивную ссылку для раздела блокнота. #### Основы Python

Как вы можете помнить, Python является объектно-ориентированным языком, кстати с неявной строгой динамической типизацией, что значит, что, во-первых, не требуется указания типа переменной при её объявлении (то есть тип переменной будет определен интерпретатором), во-вторых, переменные имеют тип и тип имеет значение при выполнении операций: не позволяется производить операции с данными разных типов, например нельзя вычесть из строки число, и в-третьих, тип переменной определяется интерпретатором непосредственно при выполнении программы (из-за динамической типизации в Python одна и та же переменная может иметь другой тип в разное время во время выполнения). 

Значит, если всё в Python является объектом, то типы данных – классы. Вы наверняка уже знакомы с большинством из встроенных типов, однако, коротко вспомним их (важно заметить, что речь идет о типах данных в третьей версии Python).

###### Числовые типы данных

Существует три различных числовых типа: целые числа (`int`), числа с плавающей запятой (`float`) и комплексные числа (`complex`). Кроме того, логические значения (`bool`) — это подтип целых чисел. Информация о точности и внутреннем представлении чисел с плавающей запятой для машины, на которой выполняется ваша программа, доступна в `sys.float_info`.

```python
import sys

sys.float_info
```

У комплексных чисел есть действительная и мнимая части, каждое из которых является числом с плавающей запятой. 

```python
z = complex(1, 2)

print("z =", z)
print("The real part of z =", z.real)
print("The imaginary part of z =", z.imag)
```

Стандартная библиотека также включает дополнительные числовые типы: `fractions.Fraction` для рациональных чисел и `decimal.Decimal` для чисел с плавающей точкой с определяемой пользователем точностью.

```python
import fractions

fraction = fractions.Fraction(1, 3)
print("fraction =", fraction)
print("float(fraction) =", float(fraction))

from decimal import Decimal, getcontext

getcontext().prec = 2
print("Decimal 1/7 with prec=2 =", Decimal(1) / Decimal(7))
print("Common 1/7 =", 1 / 7)
```

###### Типы последовательностей

Существует три основных типа последовательности: списки (`list`), кортежи (`tuple`) и диапазоны (`range`). 

**Списки** — изменяемые последовательности, обычно используемые для хранения массива однородных элементов. Отметим, что списки могут быть однородными, то есть содержать один и тот же тип объектов, но могут и не быть, то есть быть гетерогенными – содержащими разные типы объектов. 

```python
heterogeneous_list = [12, "hello", 3.5, [2, "world"]]
homogenous_list = ["hello", "world", "goodbye"]
print("heterogeneous_list =", heterogeneous_list)
print("homogenous_list =", homogenous_list)
slice_list = list(range(10))
print("Исходный список: ", slice_list)
print("Срез от 1 до -1 с шагом 2: ", slice_list[1:-1:2])
```

**Кортежи** — неизменяемые последовательности, обычно используемые для хранения коллекций разнородных данных. Кортежи также используются в случаях, когда требуется неизменная последовательность однородных данных (например, для хранения в `set` или `dict`). 

Обратите внимание, что на самом деле кортеж образует запятая, а не круглые скобки. Скобки необязательны, за исключением случая пустого кортежа или для избежания синтаксической двусмысленности. 

```python
_tuple = 4, 5, 6
print(type(_tuple))
```

Для разнородных коллекций данных, где доступ по имени понятнее, чем обращение по индексу, `collections.namedtuple()` может быть более подходящим выбором, чем кортеж: [collections.namedtuple (документация)](https://docs.python.org/3/library/collections.html#collections.namedtuple).

**Диапазон** представляет собой неизменяемую последовательность чисел и обычно используется для выполнения цикла `for`. Аргументы конструктора диапазона должны быть целыми числами (`range([start,] stop [[, step]])`). Если аргумент шага опущен, то по умолчанию он равен 1. Если аргумент начала опущен, то по умолчанию он равен 0. Если шаг равен нулю, возникает ошибка `ValueError`. Преимущество диапазона перед списком или кортежем состоит в том, что объект диапазона всегда будет занимать один и тот же (небольшой) объем памяти, независимо от размера диапазона, который он представляет (поскольку хранит только значения начала, конца и шага, вычисляя отдельные элементы и поддиапазоны по мере необходимости). 

```python
print("list(range(10)):", list(range(10)))
print("list(range(1, 11)):", list(range(1, 11)))
print("list(range(0, 30, 5))", list(range(0, 30, 5)))
print("list(range(0, 10, 3))", list(range(0, 10, 3)))
print("list(range(0, -10, -1))", list(range(0, -10, -1)))
print("list(range(0, 10, -1))", list(range(0, 10, -1)))
print("list(range(20, 10, -1))", list(range(20, 10, -1)))
print("list(range(0))", list(range(0)))
print("list(range(1, 0))", list(range(1, 0)))

print()

r = range(0, 20, 2)
print("r =", r)
print("11 in r =", 11 in r)
print("10 in r =", 10 in r)
print("r.index(10) =", r.index(10))
print("r[5] =", r[5])
print("r[:5] =", r[:5])
print("r[5:] =", r[5:])
print("r[-1] =", r[-1])
```

Проверка объектов диапазона на равенство с помощью `==` и `!=` сравнивает их как последовательности. То есть два объекта диапазона считаются равными, если они представляют одну и ту же последовательность значений. Обратите внимание, что два объекта диапазона, которые считаются равными, могут иметь разные атрибуты начала, конца и шага.

```python
print(range(0) == range(2, 1, 3))
print(range(0, 3, 2) == range(0, 4, 2))
```

Текстовые данные в Python обрабатываются с помощью объектов строк (`str`). 

**Строки** — неизменяемые последовательности кодовых точек Unicode. Кодовая точка — это целочисленное значение, однозначно идентифицирующее данный символ. 

Строки в тройных кавычках могут занимать несколько строк.

Строковые литералы, которые являются частью одного выражения и имеют только пробелы между ними, будут неявно преобразованы в один строковый литерал.

Передача байтового объекта в конструктор `str()` без аргументов кодирования возвращает неформальное строковое представление. В рамках курса предполагается, что вы знакомы с кодировкой символов в Python, так что просто отметим, что в третьей версии по умолчанию используется кодировка UTF-8. Дополнительные сведения о классе str, его методах и различных кодировках смотрите в документации: [Исчерпывающее руководство по Юникоду и кодировке символов в Python](https://tproger.ru/translations/unicode-and-encoding-python-guide/).

```python
str_without_enc = str(b'Hello!')
str_with_enc = str(b'Hello!', encoding='utf-8')
print("str_without_enc:", str_without_enc)
print("str_with_enc:", str_with_enc)

unicode_str = u'Пример'
print("unicode_str:", unicode_str, type(unicode_str))
encoded_str = unicode_str.encode('utf-8')
print("encoded_str:", encoded_str, type(encoded_str))
decoded_unicode_str = encoded_str.decode('utf-8')
print("decoded_unicode_str:", decoded_unicode_str, type(decoded_unicode_str))
decoded_cp_str = encoded_str.decode('cp1251')
print("decoded_cp_str:", decoded_cp_str, type(decoded_cp_str))
```

###### Типы наборов

**Множество** (`set`, `frozenset`) — это неупорядоченный набор хешируемых объектов. Наборы часто используются для удаления дубликатов из последовательностей или различных математических операций с множествами. Будучи неупорядоченной коллекцией, наборы не сохраняют положение элемента или порядок вставки. Соответственно, наборы не поддерживают, например, индексацию или срезы. Существует два типа наборов: `set` и `frozenset`. 

Тип `set` является изменяемым, а значит не имеет хэш-значения и не может использоваться как ключ словаря. 

```python
# Set
first_list = [1, 2, 3, 4, 2]
first_set = set(first_list)
first_set.add(5)
second_set = {1, 2, 1}
second_set.add(6)
second_set.remove(2)
print("first_set = ", first_set)
print("second_set = ", second_set)
print("first_set & second_set =", first_set & second_set)
print("first_set | second_set =", first_set | second_set)
print("first_set - second_set =", first_set - second_set)
print("second_set - first_set =", second_set - first_set)
print("first_set ^ second_set =", first_set ^ second_set)  # Симметрическая разность
print("second_set ^ first_set =", second_set ^ first_set)  # Симметрическая разность
```

Тип `frozenset` является неизменяемым и хешируемым - его содержимое не может быть изменено после его создания.

```python
frozen_set = frozenset([1, 2])
# frozen_set.add(3)
# frozen_set.remove(1)
```

###### Типы сопоставления (mapping)

Маппинг — определение соответствия данных между последовательностями элементов. Объект сопоставления отображает хешируемые значения на произвольные объекты. **Сопоставления** — это изменяемые объекты. В настоящее время в третьей версии Python существует только один стандартный тип сопоставления – словарь (`dict`). Значения, которые не могут быть хешированы, то есть значения, содержащие изменяемые объекты (которые сравниваются по значению, а не по идентификатору объекта), не могут использоваться в качестве ключей. Числовые типы, используемые в качестве ключей, подчиняются обычным правилам сравнения: если два числа считаются равными (например, 1 и 1.0), то они могут использоваться взаимозаменяемо для индексации одного и того же значения в словаре. Обратите внимание, поскольку компьютеры хранят числа с плавающей запятой в виде приближения, чаще всего неразумно использовать их в качестве ключей словаря. 

```python
example_dict = {1: 2}
print(example_dict)
example_dict[1.0] = 3
print(example_dict)

# примеры создания dict
a = dict(one=1, two=2, three=3)
b = {'one': 1, 'two': 2, 'three': 3}
c = dict(zip(['one', 'two', 'three'], [1, 2, 3]))
d = dict([('two', 2), ('one', 1), ('three', 3)])
e = dict({'three': 3, 'one': 1, 'two': 2})
f = dict({'one': 1, 'three': 3}, two=2)
print(a == b == c == d == e == f)
```

###### NoneType

**`None`** - единственная переменная типа `NoneType`. `None` часто возникает в результате выполнения функций (которые ничего не возвращают), также используется для обозначения отсутствия значения, например, когда аргументы по умолчанию не передаются в функцию. 

```python
print(type(None))
print(type(print()))
```

Формально есть два способа проверить равенство `None`: с помощью ключевого слова `is` и с помощью `==`. Однако, проверка на равенство `None` с использованием `==` работает корректно только со встроенными типами данных, а также с пользовательскими классами, в которых корректно реализован метод сравнения (`__eq__`).

```python
class Foo(object):
    def __eq__(self, other):
        return True

f = Foo()
print(f == None)
print(f is None)
```

###### IF

Перейдем к освежению воспоминаний о синтаксисе языка Python. Пожалуй, наиболее известным оператором является условный оператор `if`. Может быть ноль или более `elif`, а `else` является необязательным. Ключевое слово `elif` является сокращением от `else if` и полезно для того, чтобы избежать чрезмерного отступа. Последовательность `if … elif … elif …` — это замена операторов `switch` или `case` в других языках.

```python
x = int(input("Введите целое число: "))

if x < 0:
    x = 0
    print('Меньше нуля заменим на ноль')
elif x == 0:
    print('Ноль')
elif x == 1:
    print('Один')
else:
    print('Много')
```

###### FOR

Оператор `for` в Python немного отличается от того, к чему вы, возможно, привыкли в других языках. Вместо того, чтобы перебирать арифметическую последовательность чисел или давать пользователю возможность определять как шаг итерации, так и условие остановки, оператор `for` в Python выполняет итерацию по элементам любой последовательности (например, списка или строки) в том порядке, в котором они находятся в последовательности. 

```python
words = ['cat', 'dog', 'chicken', 'python']
for word in words:
    print(word, len(word))
```

Код, который изменяет коллекцию при повторении той же самой коллекции, может быть сложно понять правильно. Вместо этого обычно проще перебрать копию коллекции или создать новую коллекцию.

```python
# Итерирование по копии
users = {"Tom": "inactive", "Pete": "active", "Kate":"active"}
print("Initial collection:", users)
for user, status in users.copy().items():
    if status == 'inactive':
        del users[user]

print("Collection after for loop:", users)

# Создание новой коллекции
users = {"Tom": "inactive", "Pete": "active", "Kate":"active"}
active_users = {}
for user, status in users.items():
    if status == 'active':
        active_users[user] = status

print("\nCollection:", users)
print("New collection:", active_users)
```

Если вам нужно перебрать последовательность чисел, вам пригодится встроенная функция `range()`, которая, как вы уже знаете, генерирует арифметические прогрессии (диапазоны). Конечная точка диапазона никогда не является частью сгенерированной последовательности. Чтобы перебирать индексы последовательности, вы можете комбинировать функции `range()` и `len()`. Повторим, что хотя во многих отношениях объект, возвращаемый функцией `range()`, ведет себя так, как будто это список - на самом деле это не так. Это объект, который возвращает последовательные элементы желаемой последовательности при итерации по нему, но на самом деле он не является списком, тем самым экономя место. Такой объект является итеративным, то есть подходит в качестве аргумента для функций и конструкций, которые ожидают чего-то, от чего могут получать последовательные элементы. Вы уже видели, что оператор `for` является как раз такой конструкцией, в то время как примером функции можно назвать `sum()`.

```python
for i in range(5):  # range(5), range(5, 10), range(0, 10, 3), range(-10, -100, -30)
    print(i)
```

```python
words = ['Mary', 'had', 'a', 'little', 'lamb']
for i in range(len(words)):
    print(i, words[i])
```

```python
sum(range(5))  # 0 + 1 + 2 + 3 + 4
```

```python
set(range(13))
```

При итерировании по последовательности индекс позиции и соответствующее значение могут быть получены одновременно с помощью функции `enumerate()`. 

```python
for index, value in enumerate(['one', 'two', 'three']):
    print(index, value)
```

Для одновременного перебора двух или более последовательностей записи могут быть объединены с помощью функции `zip()`. 

```python
questions = ['name', 'quest', 'favorite color']
answers = ['lancelot', 'the holy grail', 'blue']
for q, a in zip(questions, answers):
    print('What is your {0}?  It is {1}.'.format(q, a))
```

Чтобы перебрать последовательность в обратном порядке, вызовите функцию `reversed()`. 

```python
for i in reversed(range(1, 10, 2)):
    print(i)
```

Чтобы перебрать последовательность в отсортированном порядке, используйте функцию `sorted()`, которая возвращает новый отсортированный список, оставляя исходный неизменным.

```python
basket = ['apple', 'orange', 'apple', 'pear', 'orange', 'banana']
for i in sorted(basket):
    print(i)
```

```python
for f in sorted(set(basket)):
     print(f)
```

###### WHILE

Оператор `while` используется для повторного выполнения, пока выражение истинно, то есть многократно проверяет выражение и, если оно истинно, выполняет первое условие; если выражение ложно (что может быть даже при первой проверке условия), выполняется `else`, если присутствует, и цикл завершается.

```python
a, b = 0, 1
while a < 10:
    print(a)
    a, b = b, a + b
else:
    print("THE END")
```

###### break, continue, pass, else

Оператор `break` принудительно выходит за пределы циклов `for` или `while`. Операторы цикла могут иметь условие `else`, которое выполняется, когда цикл завершается, но не когда цикл завершается оператором `break`. 

```python
for n in range(2, 10):
    for x in range(2, n):
        if n % x == 0:
            print(n, 'равно', x, '*', n // x)
            break
    else:
        print(n, '- простое число')
```

Оператор `continue` продолжает следующую итерацию цикла. 

```python
for n in range(2, 10):
    if n % 2 == 0:
        print("Четное:", n)
        continue
    print("Нечетное:", n)
```

Оператор `pass` ничего не делает. Его можно использовать, когда оператор требуется синтаксически, но программа не требует действий, например, для создания классов или в качестве наполнения функции, когда вы работаете над новым кодом, что позволяет вам думать на более абстрактном уровне.

```python
class Class:
    pass

def function(*args):
    pass
```

###### Syntax sugar

Напоследок напомним некоторые синтаксические возможности, так называемый «синтаксический сахар». Например, все вышеназванные циклы и условия могут быть записаны в одну строку. 

```python
# one line
x = 1
a = 2 if x > 1 else 3
print(a)

lst = [1, 2, 3, 4]
doubled = [num * 2 for num in lst]
print(doubled)

while a > 0: a = a - 1; print(a)
```

C помощью ключевого слова `lambda` можно создавать небольшие анонимные функции. Лямбда-функции можно использовать везде, где требуются функциональные объекты, но они синтаксически ограничены одним выражением. Семантически это просто синтаксический сахар для определения обычной функции. 

```python
# lambda arguments: expression
lambda_function = lambda x: x * x
print(type(lambda_function))
print(lambda_function(5))
print(list(map(lambda x: x ** x, range(1, 6, 2))))
print(list(map(lambda x, y: x * y, range(1, 6, 2), range(2, 7, 2))))
```

Наконец, важно упомянуть встроенную функцию `map()`, возвращающую итератор, применяющий функцию к каждому элементу итерации.

```python
# map(function, iterable, [iterable 2, iterable 3, ...])
from math import sqrt, pow
list_1 = [1, 2, 3]
print(map(sqrt, list_1))
print(list(map(sqrt, list_1)))
list_2 = [3, 4, 5, 6]
print(list(map(pow, list_1, list_2)))
print(list(map(pow, list_2, list_1)))
```

###### Numpy

Познакомимся с основными библиотеками Python, которые потребуются нам для работы. Начнем с NumPy. Разберемся как применять общие функции к n-мерным массивам (без использования циклов for), и поймём свойства оси и формы для n-мерных массивов. Но начнем по порядку.

```python
import numpy as np
```

Основной объект NumPy - однородный многомерный массив. По сути, это таблица элементов (обычно чисел) одного типа, индексированная набором неотрицательных целых чисел. В NumPy измерения называются осями. Например, координаты точки в трехмерном пространстве \[1, 2, 1\] имеют одну ось. Эта ось имеет 3 элемента, поэтому мы говорим, что она имеет длину 3. В примере, массив имеет 2 оси. Первая ось имеет длину 2, вторая ось имеет длину 3. Класс массива NumPy называется ndarray.

```python
# numpy.ndarray
np.array([[ 1., 0., 0.],
          [ 0., 1., 2.]])
```

Существует несколько способов создания массивов. Например, вы можете создать массив из обычного списка или кортежа Python, используя функцию массива. Тип результирующего массива определяется типом элементов в последовательностях. 

```python
a = np.arange(15).reshape(3, 5)
a
```

```python
print(a.shape)
print(a.ndim)
print(a.size)
```

```python
b = np.array([6, 7, 8.5])
print(a.dtype)
print(b.dtype)
```

Распространенная ошибка заключается в вызове массива с несколькими аргументами, а не в предоставлении одной последовательности в качестве аргумента. 

```python
a = np.array(1, 2, 3, 4)    # WRONG
a = np.array([1, 2, 3, 4])  # RIGHT
```

`array()` преобразует последовательности последовательностей в двумерные массивы, последовательности последовательностей последовательностей в трехмерные массивы и так далее.

```python
m = np.array([(1.5, 2, 3), (4, 5, 6)])
m
```

Особенности обращения по индексу к спискам и NumPy-массивам:

```python
l = list(m)
print(l[0][2])
print(m[0][2])
print(m[0, 2])
```

Часто бывает, что элементы массива изначально неизвестны, но известен его размер. Для этого NumPy предлагает несколько функций для создания массивов с начальным содержимым заполнителя, что сводит к минимуму необходимость увеличения массивов (что является дорогостоящей операцией). 

Функция `zeros` создает массив, полный нулей:

```python
np.zeros((3, 4))
```

Функция `one` создает массив, полный единиц:

```python
np.ones((2, 3, 4))
```

Функция `empty` создает массив, начальное содержимое которого является случайным и зависит от состояния памяти. 

```python
np.empty((2, 3))
```

По умолчанию dtype созданного массива - float64.

Для создания последовательностей чисел NumPy предоставляет функцию `arange`, которая аналогична встроенному диапазону (`range`), но возвращает массив. 

```python
np.arange(0, 2, 0.2)  # it accepts float arguments
```

Кроме того, можно использовать функцию `linspace`, которая получает в качестве аргумента количество элементов, которое мы хотим получить, вместо шага.

```python
from numpy import pi
np.linspace(0, 2 * pi, 10)  # note that this is a segment, not a half-interval
```

Обратите внимание, что, если массив слишком велик для печати, NumPy автоматически пропускает центральную часть массива и печатает только «углы». 

```python
print(np.arange(10000).reshape(100,100))
```

Чтобы отключить это поведение и заставить NumPy печатать весь массив, вы можете изменить параметры вывода с помощью `set_printoptions`.

```python
import sys
np.set_printoptions(threshold=sys.maxsize)
print(np.arange(10000).reshape(100,100))
```

###### Базовые операции

Арифметические операции с массивами применяются поэлементно. В их результате создается новый массив и заполняется результатом. Однако, некоторые операции, такие как `+=` и `*=`, изменяют существующий массив вместо того, чтобы создавать новый. Оператор произведения `*` тоже работает поэлементно в массивах NumPy. 

```python
a = np.array([[20, 30], [40, 50]])
b = np.arange(4).reshape(2, 2)
print(a - b)
print(b ** 2)
print(10 * np.sin(a))
print(a < 35)
```

```python
l = [1, 2, 3]
print(l * 2)
print(l ** 2)
```

Матричное произведение может быть выполнено с помощью оператора `@` (в python> = 3.5) или с использованием функции или метода `dot()`. 

```python
print(a * b)
print(a @ b)
print(a.dot(b))
print(np.dot(a, b))
```

Многие унарные операции, такие как вычисление суммы всех элементов в массиве, реализованы как методы класса `ndarray`. По умолчанию эти операции применяются к массиву, как если бы он был списком чисел, независимо от его формы. 

```python
print(a.sum())
print(a.min())
print(a.max())
```

Однако, указав параметр оси, вы можете применить операцию вдоль указанной оси массива. 

```python
print(b.sum(axis=0))     # sum of each column
print(b.min(axis=1))     # min of each row
print(b.cumsum(axis=1))  # cumulative sum along each row
```

NumPy предоставляет знакомые математические функции, такие как `sin`, `cos` и `exp`. В NumPy они называются «универсальными функциями». Аналогичным образом, в NumPy эти функции работают поэлементно с массивом, создавая массив в качестве вывода.

```python
print(np.exp(b))
print(np.sqrt(b))
```

Одномерные массивы можно индексировать, нарезать и итерировать, как встроенные типы последовательностей Python (например, списки). 

```python
a = np.arange(10) ** 3
print(a)
a[:6:2] = 1000
print(a)
```

Многомерные массивы могут иметь по одному индексу для каждой оси. Эти индексы даются в виде кортежа через запятую. 

```python
def f(x, y):
     return 2 * x + y

b = np.fromfunction(f, (3, 4))
print(b)
print(b[0, 1])
print(b[1:3, 1])  # two rows in the second column of b
```

Когда предоставляется меньше индексов, чем количество осей, отсутствующие индексы считаются полными срезами.

```python
print(b[2])
```

Итерация по многомерным массивам выполняется относительно первой оси. 

```python
for row in b:
     print(row)
```

Однако, если вы хотите выполнить операцию с каждым элементом в массиве, можно использовать атрибут `flat`, который позволяет итерировать по всем элементам массива. 

```python
for element in b.flat:
     print(element)
```

###### Манипуляции с формой

Форма массива определяется количеством элементов по каждой оси. Однако, форму массива можно изменить с помощью различных команд. 

```python
a = np.floor(10 * np.random.rand(3, 4))
print(a)
print(a.shape)
```

Обратите внимание, что следующие три команды возвращают измененный массив, но не изменяют исходный массив. 

```python
print(a.ravel())       # returns the array, flattened
print(a.reshape(6, 2)) # returns the array with a modified shape
print(a.T)             # returns the array, transposed
```

Но функция `reshape` уже возвращает свой аргумент с измененной формой, тогда как метод `ndarray.resize` изменяет сам массив. 

```python
a.resize((2, 6))
a
```

Если размер задается как `-1` в операции изменения формы (`reshape`), то он вычисляется автоматически.

```python
print(a.reshape(3, -1))
```

Массивы можно объединять вдоль осей. 

```python
a = np.floor(10 * np.random.rand(3, 4))
b = np.floor(10 * np.random.rand(3, 4))
print(a)
print(b)
print(np.vstack((a, b)))
print(np.hstack((a, b)))
```

Функция `column_stack` складывает одномерные массивы как столбцы в двумерный массив (эквивалентно `hstack` для 2D-массивов). С другой стороны, функция `row_stack` эквивалентна `vstack` для любых входных массивов. Для массивов с более чем двумя измерениями `hstack` и `vstack` допускают необязательные аргументы, указывающие номер оси, вдоль которой должно происходить объединение.

```python
a = np.array([1, 2, 3])
b = np.array([1, 2, 3])
print(np.vstack((a, b)))
print(np.row_stack((a, b)))
print(np.hstack((a, b)))
print(np.column_stack((a, b)))
```
